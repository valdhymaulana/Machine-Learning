```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

# 1. Sesuaikan path direktori sesuai screenshot (masuk ke dalam sub-folder)
DATASET_DIR = 'dataset_sampah/TrashType_Image_Dataset'

# 2. Preprocessing & Data Augmentation (Slide 25) + Validation Split
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    zoom_range=0.2,
    horizontal_flip=True,
    validation_split=0.2  # Kunci untuk membagi data otomatis (80% Train, 20% Val)
)

# 3. Memuat Data Training
print("Memuat Data Training:")
train_data = train_datagen.flow_from_directory(
    DATASET_DIR,
    target_size=(128, 128),
    class_mode='categorical',
    subset='training'
)

# 4. Memuat Data Validasi
print("Memuat Data Validasi:")
val_data = train_datagen.flow_from_directory(
    DATASET_DIR,
    target_size=(128, 128),
    class_mode='categorical',
    subset='validation'
)

# 5. Membangun Arsitektur CNN (Persis Sesuai Slide 25)
model = Sequential([
    Conv2D(32, (3,3), activation='relu', input_shape=(128, 128, 3)),
    MaxPooling2D(2,2),
    Conv2D(64, (3,3), activation='relu'),
    MaxPooling2D(2,2),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(train_data.num_classes, activation='softmax')
])

# 6. Kompilasi Model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# 7. Eksekusi Training
print("\n--- Memulai Proses Training ---")
history = model.fit(
    train_data,
    validation_data=val_data, # Memasukkan data uji
    epochs=10
)

# 8. Menyimpan Matriks Bobot (Sangat Krusial!)
model.save('model_sampah_kampus2.keras')
print("\nTraining Selesai! Model berhasil diamankan sebagai 'model_sampah_kampus.h5'.")
```
